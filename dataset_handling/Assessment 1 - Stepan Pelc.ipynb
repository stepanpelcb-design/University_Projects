```python
#Student Name: Stepan Pelc
#MISIS: M01087679

#Finding a breach in week 4 html file in: 5.2. Quantiles
print(f"In the week 4 html file was Python style guide PEP-0008 breached in this line of code:\n\"diamonds.price.quantile([0.2,0.4])\"\nAccording to PEP 8, missing a space after a comma is incorrect. So there is a breach in \"[0.2,0.4]\".\nThe correct way of writing this code line is:\n\"diamonds.price.quantile([0.2, 0.4])\"")


#from fontTools.merge.util import first

#Question 1

# a)

print(f"\nQuestion 1, a)\n")

import pandas as pd

#Folder path of all data
folder_path = "/Users/stepanpelc/Library/CloudStorage/OneDrive-MiddlesexUniversity/Lessons/1 Year/Introduction to Artificial Intelligence and Data Science/Assessment 1/Metropolitan_Police_Station/"

#creating list of each month in the timeframe
months = pd.date_range(start='2023-08-01', end='2025-08-01', freq='ME').strftime('%Y-%m')
months = list(months)

#creating empty list for all data
df = []

#loop for reading each month data to the DataFrame
for month in months:
    df.append(pd.read_csv(f"{folder_path}{month}/{month}-metropolitan-street.csv"))
#concatinating list with all data for each month into a dataframe
df = pd.concat(df)
#cleaning the data and printing info and the first 5 lines of data
df.isnull().sum()
#printing info about the df
print(df.info())
print(df.head())

#creating list for all columns I want to keep
columns = ["Month", "Location", "LSOA name", "Crime type"]
#defining new dataframe with removed variables
df = df[columns]
#printing info about the changed df
print(df.info())
print(df.head())

# b)

print(f"\nQuestion 1, b)\n")

#all unique crime types
crime_types = df["Crime type"].unique()


#creating a dataframe for crime type Anti-social behaviour only
antisocial_behaviour = df[df["Crime type"] == "Anti-social behaviour"]

#counting how many times crime happened in each London Borough
count = antisocial_behaviour["LSOA name"].value_counts()

#finding Borough with the most crimes
max_borough = count.idxmax()
#printing
print(f"The most \"Anti-social behaviour\" incidents happened in {max_borough}.")

#all bicycle theft crimes from the whole timeframe
bicycle_theft = df[df["Crime type"] == "Bicycle theft"]
#narowing dataframe only for June 2024
bicycle_theft = bicycle_theft[bicycle_theft["Month"] == "2024-06"]
#counting how many times the crime happened in each London borough
count = bicycle_theft["LSOA name"].value_counts()
#finding in which borough the crime happened the most
max_borough = count.idxmax()
#printing
print(f"The most \"Bicycle theft\" crimes in June 2024 happened in {max_borough}.")

# c)

print(f"\nQuestion 1, c)\n")

#importing lets_plot
from lets_plot import *
#setting up html
LetsPlot.setup_html()

print("The first graph:")

#Graph number 1

#Count of all the crimes from the start date to the end date

graph1 = (
    ggplot(data=df, mapping=aes(x="Crime type", fill="Crime type")) + #chosing dataframe, and only crime type for x aes. The y aes is automatically showing the count of that crime type.
    geom_histogram(show_legend=None) + #creating histogram
    theme(legend_position="none") + #chosing not to show the legend
    labs(title="The number of times each crime happened") #creating the title of the graph
)

#displaying the graph
display(graph1)

#This graph helps us to see how many times each crime type happened throughout the whole timeframe.


print("The second graph:")
#Graph number 2

#creating a new row variable for counting the number of crimes each month
df["Total Crime Count Per Month"] = df.groupby("Month")["Crime type"].transform("count")

#how the total number of crimes changed over time
graph2 = (
ggplot(data=df, mapping=aes(x="Month", y="Total Crime Count Per Month", color="Total Crime Count Per Month")) + #chosing which data to show on each aes
geom_line(alpha=1, size=3) + #creating a line graph and setting the thickness of the line to 3
labs(title="Total number of each crime type") #choosing the title of the graph
)
display(graph2) #displaying the second graph

#This graph shows the total number of crimes each month and how the number of crimes changed over time. It shows in which month the crimes happened the most and the least.

print("The third graph:")
#Graph number 3

#total number of crimes, and crime types for each month

graph3 = (
ggplot(data=df, mapping=aes(x="Month", y="Total Crime Count Per Month", fill= "Crime type")) + #chosing which data to show on each aes and filling colour of bins with different crime types
geom_bar(alpha=1) + #creating a histogram
labs(title="Crimes each month") #title of the graph
)
display(graph3) #displaying the third graph

#This graph shows the number of times all the crimes happened each month, and also it shows the number of each crime type that month with different color. With that it is pretty easy to see what proportion each crime represents that month.



# f)
#importing numpy
import numpy as np

print(f"\nQuestion 1, f)\n")

#calculating total number of crimes, that happened at each location
location_crime_count = df.groupby("Location")["Crime type"].count().reset_index(name="Total Crime Count")

#narowing the df to only locations between 200 and 500 total crimes
location_crime_count = location_crime_count[(location_crime_count["Total Crime Count"] >= 200) & (location_crime_count["Total Crime Count"] <= 500)]

#choosing the first location from all the locations
location = location_crime_count.iloc[0]["Location"]

#printing that one location
print(f"Location in range between 200 and 500: {location}")

#All the crimes that happened at our location
crimes_at_location = df[df["Location"] == location]

#summing the same crime types at our location
each_crime = crimes_at_location["Crime type"].value_counts()
print(f"Number of crimes at the location: {each_crime}")

#the total number of crimes
all_crimes = len(crimes_at_location)

#the total number of other theft crimes
other_theft = crimes_at_location[crimes_at_location["Crime type"] == "Other theft"].value_counts().sum()

#calculating the proportion of total crimes
proportion_other_theft = other_theft / all_crimes

#calculating standard error for the other theft proportion
SE = np.sqrt(proportion_other_theft * (1 - proportion_other_theft) / all_crimes)

#calculating the 95% confidence intervals
CI95_plus = proportion_other_theft + (1.96 * SE)
CI95_minus = proportion_other_theft - (1.96 * SE)

print(f"I am 95% confident, that crime type \"Other theft\" happened between: {CI95_plus * 100:.1f}% and {CI95_minus * 100:.1f}% times of all the {all_crimes} crimes.")



#Question 2
print(f"\nQuestion 2\n\n")

knowledge = [
    ("Alan Turing", "isa", "Person"),
    ("Alan Turing", "field", "Computer Science"),
    ("Computer Science", "SubfieldOf", "STEM"),
    ("Machine Learning", "SubfieldOf", "Artificial Intelligence"),
    ("Artificial Intelligence", "SubfieldOf", "Computer Science"),
    ("STEM", "SubfieldOf", "Academia"),
    ("LLM", "SubfieldOf", "Artificial Intelligence")
]
print(f"Knowledge: {knowledge}")

# a)

print(f"\nQuestion 2, a)\n")

print("This knowledge is represented as a semantic net, because there are Nodes (\"Alan Turing\", as an object) and Arcs (\"isa\", as a relationship). The arcs connect nodes. \n In Semantic nets, the relationships are very clear and easily visible, but sometimes the reasoning can be difficult. The system must search across multiple connections because there is no direct connection between the two nodes. Throughout Semantic nets, we can see how the knowledge is connected, and we can use search algorithms to help us find the shortest and fastest path to our goal. \nWe can use, for instance, Forward Chaining, to find our goal. Forward chaining starts with the current knowledge, and it applies rules on its path to the goal to discover new facts until it reaches the goal. Another search algorithm is Depth-first Search. This search algorithm explores every possible path until it reaches the goal. It also follows some rules, which we give, to guide it on how to search. Then it will go as deep as it can on one path and use backtracking to go back if there are no more states to explore on that path.\nThis knowledge representation is, in my opinion, great for highlighting relationships between nodes. And we can easily see, what each node is a part of. The main limitation of this knowledge representation is that there is no direct connection between some nodes. For example between \"Alan Turing\" and \"Artificial Intelligence\", we would need to search that through all the other nodes and arcs.")

# b)

print(f"\nQuestion 2, b)\n")

#new knowledge
new_knowledge = [
    ("Alan Turing", "field", "Mathematics"),
    ("Pythagoras", "field", "Mathematics"),
    ("Pythagoras", "isa", "Person"),
    ("Mathematics", "SubfieldOf", "STEM"),
    ("Data Science", "SubfieldOf", "Computer Science"),
    ("Deep Learning", "SubfieldOf", "Machine Learning"),
    ("Engineering", "SubfieldOf", "STEM"),
    ("Programming Languages", "SubfieldOf", "Computer Science"),
    ("ChatGPT", "isa", "LLM"),
    ("Python", "SubfieldOf", "Programming Languages"),
]

#adding new knowledge to previous knowledge
for s, p, o in new_knowledge:
    knowledge.append((s, p, o))

print(f"The knowledge with new added triplets: {knowledge}")

# c)

print(f"\nQuestion 2, c)\n")

#creating the query function
def query(knowledge, subject=None, predicate=None, object=None):
    """Returns all triples from the knowledge that match the subject."""
    #creating empty list to store matched variable
    triple = []
    #loop for checking each triple in the knowledge
    for s, p, o in knowledge:
        if (subject is None or s == subject) and (predicate is None or p == predicate) and (object is None or o == object):
            triple.append((s, p, o))
    return triple

#printing the query
print(f"Knowledge with subject \"Alan Turing\": {query(knowledge, subject = "Alan Turing")}")
print(f"Knowledge with predicate \"SubfieldOf\": {query(knowledge, predicate = "SubfieldOf")}")
print(f"Knowledge with object \"Artificial Intelligence\": {query(knowledge, object = "Artificial Intelligence")}")


#Question 3

# a)

print(f"\nQuestion 3, a)\n")

def perceive(x):
    """Returns an output of perceptron from input values"""
    #defining weights
    weights = ([0.2, 0.4, 0.3, 0.6, 0.2])
    #creating variable for output
    output = 0
    #loop for multiplying weights with input nodes
    for number in range(len(weights)):
        #adding multiplied value to the output variable
        output += weights[number] * x[number]
    #checking if the output is less then 1
    if output >= 1:
        return 1
    else:
        return 0


#creating example arrays
array1 = ([0.2, 0.3, 0.1, 0.5, 0.2])
array2 = ([0.5, 0.5, 0.1, 1, 1])
array3 = ([0.2, 0.3, 0.1, 0.5, 1])

print(f"The input nodes {array1} will return: {perceive(array1)}")
print(f"The input nodes {array2} will return: {perceive(array2)}")
print(f"The input nodes {array3} will return: {perceive(array3)}")


# b)

print(f"\nQuestion 3, b)\n")

print("This knowledge representation is called a network of neurons. This knowledge consist of multiple neurons arranged after each other in layers. The thing, which makes this knowledge special is that each neuron have its own weights. When that neuron receives an input, the neuron multiplies the input with its weight, and if the output is greater (or equal to) a value (which we define by a function), the neuron will fire. The knowledge is basically represented by all the neurons, and there weights.\nWe simply feed data and train the algorithm with this data repeatedly. Of course the algorithm will make mistakes, but it will learn from them and make necessary adjustments to get more accurate. That is how we acquire knowledge for this knowledge representation.")
```
